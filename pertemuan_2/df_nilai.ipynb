```python
import pandas as pd

data = {
    'nama': ['Hasan', 'Husein', 'Tiut', 'Titut', 'Asep'],
    'uts': [75, 85, 90, 70, 88],
    'uas': [88, 78, 90, 86, 100]
}

df_nilai = pd.DataFrame(data)
df_nilai['nilai_akhir'] = (df_nilai['uts'] * 0.4) + (df_nilai['uas'] * 0.6)

df_nilai
```

```python
df_nilai_di_atas_80 = df_nilai[df_nilai['nilai_akhir'] > 80]
df_nilai_di_atas_80
```
