```python
from pymongo import MongoClient

# Membuat koneksi ke MongoDB yang berjalan di localhost, port default 27017
client = MongoClient("mongodb://localhost:27017/")

# Memilih (atau otomatis membuat) database
db = client["praktikum_bigdata"]

# Memilih (atau otomatis membuat) collection
koleksi_produk = db["produk"]

print("Koneksi ke MongoDB berhasil!")
print("Daftar database yang ada:", client.list_database_names())
```

```python
koleksi_produk.insert_many([
    {"nama_produk": "Keripik Singkong", "kategori": "Snack", "harga": 12000, "stok": 70, "tags": ["gurih", "renyah"]},
    {"nama_produk": "Cokelat Batang", "kategori": "Snack", "harga": 18000, "stok": 55, "tags": ["manis"]},
    {"nama_produk": "Kacang Panggang", "kategori": "Snack", "harga": 10000, "stok": 90, "tags": ["gurih", "murah"]},
    {"nama_produk": "Biskuit Cokelat", "kategori": "Snack", "harga": 9000, "stok": 100, "tags": ["manis", "murah"]},
    {"nama_produk": "Popcorn Karamel", "kategori": "Snack", "harga": 22000, "stok": 40, "tags": ["manis", "best-seller"]},
])

print("Jumlah dokumen di collection 'produk':", koleksi_produk.count_documents({}))
```

```python
print("Produk dengan harga < 20000:")
for p in koleksi_produk.find({"harga": {"$lt": 20000}}):
    print(f"  - {p['nama_produk']} (Rp{p['harga']:,})")
```

```python
hasil = koleksi_produk.update_many({}, {"$set": {"diskon": 0}})

print("Dokumen yang diperbarui:", hasil.modified_count)
for p in koleksi_produk.find({}, {"_id": 0, "nama_produk": 1, "diskon": 1}):
    print(f"  - {p['nama_produk']}: diskon={p['diskon']}")
```

MongoDB lebih tepat dipakai untuk menyimpan rekam medis pasien di rumah sakit. Setiap pasien memiliki data yang berbeda-beda: sebagian punya riwayat alergi, hasil lab, atau catatan operasi, sementara yang lain tidak, dan jenis pemeriksaannya terus bertambah. Dengan *flexible schema*, tiap dokumen pasien bisa memiliki field berbeda tanpa mengubah struktur keseluruhan, sedangkan pada tabel relasional hal ini menghasilkan banyak kolom kosong atau memerlukan banyak tabel tambahan.
